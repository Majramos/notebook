# Convert between hex and RGB

## Hex to RGB
- Converting a hexadecimal color code to a tuple of integers corresponding to its RGB components
- use python built in function [int()](https://docs.python.org/3/library/functions.html#int) with base 16 `int(string, base=16)`

```python
def hex_to_rgb(hex):
    h = hex.lstrip("#")
    return tuple(int(h[i:i+2], 16) for i in (0, 2, 4))

print(hex_to_rgb("#c96d9d"))
print(hex_to_rgb("#fa0515"))
```

## RGB to  Hex
- Generate a zero-padded hexadecimal value using `{:02X}`
- X: [Hex format. Outputs the number in base 16, using upper-case letters for the digits above 9](https://docs.python.org/3/library/string.html#format-specification-mini-language)

```python
def rgb_to_hex(r, g, b):
  return f"{r:02X}{g:02X}{b:02X}"

print(rgb_to_hex(201, 109, 157))
print(rgb_to_hex(250, 5, 21))
```
