# Reversibly encode alphanumeric string to integer

```python
m = "test123"

# string to bytes
m_bytes = m.encode("utf-8")

# bytes to integer
m_int = int.from_bytes(m_bytes, byteorder="big")

# integer to bytes
m_bytes2 = m_int.to_bytes(((m_int.bit_length() + 7) // 8), byteorder="big")

# bytes to string
m2 = m_bytes2.decode("utf-8")

print(m == m2)
```

# Encoder Object

```python
class BytesIntEncoder:

    @staticmethod
    def encode(b: bytes) -> int:
        return int.from_bytes(b, byteorder='big')

    @staticmethod
    def decode(i: int) -> bytes:
        return i.to_bytes(((i.bit_length() + 7) // 8), byteorder='big')


s = 'Test123'
b = s.encode()
BytesIntEncoder.encode(b)
```
