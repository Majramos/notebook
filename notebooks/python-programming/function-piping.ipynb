# Function Piping

```python
import pandas as pd
```

```python
def pipe(original):
    class PipeInto:
        data = {"function": original}

        def __init__(self, *args, **kwargs):
            self.data["args"] = args
            self.data["kwargs"] = kwargs

        def __rrshift__(self, other):
            return self.data["function"](other, *self.data["args"], **self.data["kwargs"])
    return PipeInto
```

```python
@pipe
def select(df, *args):
    cols = [x for x in args]
    return df[cols]

@pipe
def rename(df, **kwargs):
    for key, val in kwargs.items():
        df = df.rename(columns={key: val})
    return df
```
