# Generate a date dimension table

```python
import pandas as pd
```

- pandas represents timestamps in nanosecond resolution, the timespan that can be represented using a 64-bit integer is limited to approximately 584 years
- [pandas timestamp limitations](https://pandas.pydata.org/pandas-docs/stable/user_guide/timeseries.html#timestamp-limitations)

```python
pd.Timestamp.min, pd.Timestamp.max
```

```python
(
    pd.DataFrame(
        pd.date_range("1900-01-01", "2200-12-31"),
        columns=["date"]
    )
    .reset_index(names="date_id")
    .assign(
        date_id=lambda x: x["date_id"]+1,
        date_number=lambda x: x["date"].dt.strftime("%Y%m%d").astype("int32"),
        year=lambda x: x["date"].dt.year,
        quarter=lambda x: x["date"].dt.quarter,
        quarter_name=lambda x: "Q"+x["quarter"].astype("string"),
        month=lambda x: x["date"].dt.month,
        month_name=lambda x: x["date"].dt.month_name(),
        month_abbr_name=lambda x: x["date"].dt.strftime("%b").str.lower(),
        day=lambda x: x["date"].dt.day,
        day_of_week=lambda x: x["date"].dt.dayofweek,
        day_name=lambda x: x["date"].dt.day_name(),
    )
)
```
