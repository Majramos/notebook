# Routing with OSRM
- routing using [Open Source Routing Machine](https://project-osrm.org/)

```python
import requests
```

## Single route travel distance and time

```python
source_lat = 38.72529995851616
source_long = -9.14996945002924

target_lat = 39.749539614456836
target_long = -8.807581271525823

waypoints = f"{source_long},{source_lat};{target_long},{target_lat}"
```

```python
ROUTE_ENDPOINT = f"https://router.project-osrm.org/route/v1/driving"

payload = {"overview": "false"}
url = f"{ROUTE_ENDPOINT}/{waypoints}"

response = requests.get(url, params=payload)
osrm_output = response.json()
osrm_output
```

### travel time (in seconds)

```python
travel_time = osrm_output.get("routes")[0].get("duration")
travel_time
```

### travel distance (in meters)
need to add the distance, in metres, from the input coordinate to the snapped coordinate

```python
travel_distance = (
    osrm_output.get("routes")[0].get("distance") + 
    osrm_output.get("waypoints")[0].get("distance") +
    osrm_output.get("waypoints")[1].get("distance")
)
travel_distance
```

## Table route 

```python
source_lat = 38.72529995851616
source_long = -9.14996945002924

target_lat1 = 39.749539614456836
target_long1 = -8.807581271525823

target_lat2 = 39.823000057018106
target_long2 = -7.493100439954117

waypoints = f"{source_long},{source_lat};{target_long1},{target_lat1};{target_long2},{target_lat2}"
waypoints
```

```python
TABLE_ENDPOINT = f"https://router.project-osrm.org/table/v1/driving"

payload = {
    "sources": "0",
    "destinations": "1;2",
    "annotations": "duration,distance"
}
url = f"{TABLE_ENDPOINT}/{waypoints}"

response = requests.get(url, params=payload)
osrm_output = response.json()
osrm_output
```
